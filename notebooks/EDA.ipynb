# MNIST Digit Recognition - Exploratory Data Analysis

This notebook explores the MNIST dataset: sample images, class distribution,
pixel intensity statistics, and sanity checks.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.data_loader import load_mnist

%matplotlib inline
sns.set_style('whitegrid')
```

## 1. Load Dataset

```python
data = load_mnist(source='tf')
x_train, y_train = data['x_train'], data['y_train']
x_test, y_test = data['x_test'], data['y_test']
print(f'Training set: {x_train.shape}, {y_train.shape}')
print(f'Test set:     {x_test.shape}, {y_test.shape}')
```

## 2. Class Distribution

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
for ax, labels, title in [(ax1, y_train, 'Training Set'), (ax2, y_test, 'Test Set')]:
    unique, counts = np.unique(labels, return_counts=True)
    ax.bar(unique, counts, color='#1A73E8')
    ax.set_xlabel('Digit')
    ax.set_ylabel('Count')
    ax.set_title(f'Class Distribution - {title}')
    ax.set_xticks(range(10))
    for i, c in zip(unique, counts):
        ax.text(i, c + 50, str(c), ha='center', fontsize=9)
plt.tight_layout()
plt.savefig('../reports/class_distribution.png', dpi=150)
plt.show()
print('Classes are roughly balanced.')
```

## 3. Sample Images per Class

```python
fig, axes = plt.subplots(10, 10, figsize=(12, 12))
for digit in range(10):
    indices = np.where(y_train == digit)[0][:10]
    for i, idx in enumerate(indices):
        axes[digit, i].imshow(x_train[idx], cmap='gray')
        axes[digit, i].axis('off')
        if i == 0:
            axes[digit, i].set_ylabel(str(digit), fontsize=14, fontweight='bold')
plt.suptitle('Sample Images per Class (rows=digit, cols=samples)', fontsize=14)
plt.tight_layout()
plt.savefig('../reports/sample_images.png', dpi=150)
plt.show()
```

## 4. Pixel Intensity Distribution

```python
fig, axes = plt.subplots(2, 5, figsize=(15, 6))
axes = axes.flatten()
for digit in range(10):
    digit_pixels = x_train[y_train == digit].ravel()
    axes[digit].hist(digit_pixels, bins=50, color='#1A73E8', alpha=0.7)
    axes[digit].set_title(f'Digit {digit}')
    axes[digit].set_xlabel('Pixel Intensity')
    axes[digit].set_ylabel('Frequency')
plt.suptitle('Pixel Intensity Distribution per Digit', fontsize=14)
plt.tight_layout()
plt.savefig('../reports/pixel_intensity.png', dpi=150)
plt.show()
```

## 5. Missing Values Check

```python
print(f'x_train NaN count: {np.isnan(x_train).sum()}')
print(f'y_train NaN count: {np.isnan(y_train).sum()}')
print(f'x_test NaN count:  {np.isnan(x_test).sum()}')
print(f'y_test NaN count:  {np.isnan(y_test).sum()}')
print('No missing values.')
```

## 6. Summary Statistics

```python
print(f'Training images:')
print(f'  Shape: {x_train.shape}')
print(f'  Data type: {x_train.dtype}')
print(f'  Pixel range: [{x_train.min()}, {x_train.max()}]')
print(f'  Mean pixel: {x_train.mean():.2f}')
print(f'  Std pixel:  {x_train.std():.2f}')
print()
print(f'Training labels:')
print(f'  Shape: {y_train.shape}')
print(f'  Unique classes: {np.unique(y_train)}')
print()
print(f'Test images:')
print(f'  Shape: {x_test.shape}')
print(f'  Pixel range: [{x_test.min()}, {x_test.max()}]')
```

## 7. Conclusion

The MNIST dataset is clean, balanced, and ready for modeling.
